# data cleaning

```python
df.isnull().sum()
```

```python
# fix missing values
df["country"].fillna("unknow",inplace=True)
```

```python
df["date_added"].fillna("na",inplace=True)
```

```python
df["cast"].fillna("NAA",inplace=True)
```

```python
df["director"].fillna("Naa",inplace=True)
```

```python
df["duration"].fillna("naa,",inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```
